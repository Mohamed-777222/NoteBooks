```python
import sqlite3
import pandas as pd
```

# Classes & Functions

```python
class DataBase:
  def __init__(self):
    self.conn = sqlite3.connect(":memory:")

  def query(self, q):
    return pd.read_sql_query(q, self.conn)

  def excute(self, q):
    self.conn.executescript(q)
    self.conn.commit()
    return "Done!"
```

```python
class library:
  def __init__(self, db):
    self.db = db

  def search(self, keyword=None, year_from=None, year_to=None):
    basic_q = """
    SELECT *
    FROM Book
    """
    conditions=[]


    if year_from :
      conditions.append(f"Year >= {year_from}")

    if year_to:
      conditions.append(f"Year <= {year_to}")

    if keyword:
      conditions.append(f"Title LIKE '%{keyword}%'")

    if keyword or year_from or year_to:
      basic_q+=" WHERE "
      basic_q+=" AND ".join(conditions)

    return self.db.query(basic_q)
```
